```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

csv_path="/kaggle/input/datasets/ashery/chexpert/train.csv"
image_root="/kaggle/input/datasets/ashery/chexpert"

df=pd.read_csv(csv_path)

diseases=[
    "Enlarged Cardiomediastinum",
    "Cardiomegaly",
    "Lung Opacity",
    "Lung Lesion",
    "Edema",
    "Consolidation",
    "Pneumonia",
    "Atelectasis",
    "Pneumothorax",
    "Pleural Effusion",
    "Pleural Other",
    "Fracture"
]

for disease in diseases:

    positives=df[df[disease]==1].head(9)

    if len(positives)==0:
        print(f"{disease}: no positive samples")
        continue

    plt.figure(figsize=(10,10))
    plt.suptitle(disease)

    for i,(_,row) in enumerate(positives.iterrows()):

        img_path=os.path.join(image_root,row["Path"])

        img=Image.open(img_path).convert("L")

        plt.subplot(3,3,i+1)
        plt.imshow(img,cmap="gray")
        plt.axis("off")

    plt.show()
```
